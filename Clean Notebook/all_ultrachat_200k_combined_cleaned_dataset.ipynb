# Impoart Libararys

```python
import pandas as pd
import glob
import os
```

# Mount Google Drive

# Define directory path

```python
folder_path = '/content/drive/MyDrive/FYP Data Cleaning/Raw Data/Data/'
output_dir = '/content/drive/MyDrive/FYP Data Cleaning/Clean Data/'
```

# Load and combine all parquet files

```python
parquet_files = glob.glob(os.path.join(folder_path, "*.parquet"))
print(f"Total Parquet files found: {len(parquet_files)}")

if len(parquet_files) == 0:
    raise FileNotFoundError("No parquet files found. Check folder_path.")

df_combined = pd.concat([pd.read_parquet(f) for f in parquet_files], ignore_index=True)

print("\n--- Combined Dataset Initial Shape ---")
print(df_combined.shape)

print("\n--- Missing Values ---")
print(df_combined.isnull().sum())
```

# Duplicate count

```python
dup_by_id     = df_combined.duplicated(subset=['prompt_id']).sum()
dup_by_prompt = df_combined.duplicated(subset=['prompt']).sum()

print("--- Duplicate Check (before cleaning) ---")
print(f"Total rows                      : {len(df_combined)}")
print(f"Duplicate rows (by prompt_id)   : {dup_by_id}")
print(f"Duplicate rows (by prompt text) : {dup_by_prompt}")
```

# Remove duplicates

```python
before = len(df_combined)
df_combined = df_combined.drop_duplicates(subset=['prompt_id']).reset_index(drop=True)

print(f"Removed   : {before - len(df_combined)}")
print(f"Remaining : {len(df_combined)}")
```

# Verify final clean shape and sample data

```python
print("\n--- Cleaned Combined Dataset Shape ---")
print(df_combined.shape)

print("\n--- Cleaned Data Sample ---")
print(df_combined.head(2))
```

# Save the single combined and cleaned dataset

```python
output_path = os.path.join(output_dir, 'all_ultrachat_200k_combined_cleaned_dataset.csv')
df_combined.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```
