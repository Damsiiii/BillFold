```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Data Cleaning/my_personal_expense_data_1.csv")
print(df.head())
```

```python
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
clean = df.copy()
```

```python
clean = clean.drop(columns=["Subcategory", "Note.1", "Account.1"])
print(clean.shape)
```

```python
clean["Date"] = pd.to_datetime(clean["Date"], format="%m/%d/%Y %H:%M")
print(clean["Date"].dtype)   # datetime64[ns]
print(clean["Date"].min(), clean["Date"].max())
```

```python
clean = clean.drop(columns=["Amount", "Currency"])
```

```python
# remove extra spaces and make lowercase
clean["Note"] = clean["Note"].str.strip().str.lower()

# 4 notes are missing: label them instead of deleting the rows
clean["Note"] = clean["Note"].fillna("unknown")
```

```python
for col in ["Account", "Category", "Income/Expense"]:
    clean[col] = clean[col].str.strip().str.lower()
```

```python
print(clean["INR"].describe())
print(clean.sort_values("INR", ascending=False).head(8)[["Date", "Category", "Note", "INR", "Income/Expense"]])
```

```python
print(pd.crosstab(clean["Category"], clean["Income/Expense"]))
```

```python
print(clean.shape)
print(clean.dtypes)
print(clean.isna().sum())
print(clean.head())

clean.to_csv("/content/drive/MyDrive/Data Cleaning/my_personal_expense_data_1_clean.csv", index=False)
```
